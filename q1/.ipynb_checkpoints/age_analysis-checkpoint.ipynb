# Age Analysis of Civilian Awards (Q1 – age part)
**Team 007 | An Analysis of India's Civilian and Military Honours**

How has the age of civilian award winners changed between the first 40 years (1954–1993) and the later period (1994–2021)?

This notebook generates the age CSVs in `age_output/` (used by the H0–H3 notebooks) and plots them.

Steps: load data → keep winners with a known age → age summaries per award/period → age bands → data coverage by year → plots.

**Needs** `region_analysis.py` and `national_awards_cleaned.csv` in the same folder (the periods, awards and loader are shared with the region analysis).

## 1. Setup

```python
import os
import pandas as pd
import plotly.express as px

from region_analysis import (AWARDS, PERIOD_1, PERIOD_2, P1_LABEL, P2_LABEL,
                             load_data, slug)

OUT_DIR = "age_output"
BINS = [0, 40, 50, 60, 70, 80, 200]
BIN_LABELS = ["<40", "40-49", "50-59", "60-69", "70-79", "80+"]
os.makedirs(OUT_DIR, exist_ok=True)
```

## 2. Load data and keep winners with a known age
Only winners whose birth year was added have an age.

```python
df = load_data()          # already has Region and Period columns
aged = df.dropna(subset=['AgeAtAward']).copy()
print(f'Winners with an age: {len(aged)} of {len(df)}')
aged.groupby(['Award', 'Period']).size().unstack(fill_value=0)
```

## 3. Age records CSV

```python
cols = ["Name", "Award", "Field", "State/Country", "Region", "Year",
        "Year of Birth", "AgeAtAward", "Period"]
aged[cols].sort_values(["Period", "Award", "Year"]).to_csv(f"{OUT_DIR}/age_records.csv", index=False)
aged[cols].head()
```

## 4. Age summaries per award and period

```python
def describe(d: pd.DataFrame) -> pd.DataFrame:
    rows = []
    for period in (P1_LABEL, P2_LABEL):
        a = d.loc[d["Period"] == period, "AgeAtAward"]
        rows.append({
            "Period": period, "n": len(a),
            "mean": round(a.mean(), 2), "median": a.median(),
            "std": round(a.std(), 2), "min": a.min(),
            "q1": a.quantile(0.25), "q3": a.quantile(0.75), "max": a.max(),
        })
    return pd.DataFrame(rows)

summaries = []
for award in AWARDS:
    sub = aged[aged["Award"] == award]
    if sub.empty:
        continue
    s = describe(sub)
    s.to_csv(f"{OUT_DIR}/{slug(award)}_age_summary.csv", index=False)
    summaries.append(s.assign(Award=award))
allsum = describe(aged)
allsum.to_csv(f"{OUT_DIR}/all_civilian_age_summary.csv", index=False)
summaries.append(allsum.assign(Award="All awards"))
age_sum = pd.concat(summaries)
age_sum
```

## 5. Age bands (10-year bands) per award and pooled

```python
aged["AgeBand"] = pd.cut(aged["AgeAtAward"], BINS, labels=BIN_LABELS, right=False)
parts = []
for award, sub in list(aged.groupby("Award")) + [("All awards", aged)]:
    t = sub.groupby(["Period", "AgeBand"], observed=False).size().reset_index(name="Count")
    t["Share_%"] = (t["Count"] / t.groupby("Period")["Count"].transform("sum") * 100).round(2)
    t.insert(0, "Award", award)
    parts.append(t)
age_bins = pd.concat(parts)
age_bins.to_csv(f"{OUT_DIR}/age_bins.csv", index=False)
age_bins[age_bins.Award == "All awards"]
```

## 6. Data coverage by year
Shows how many winners have an age each year. Ages are missing for most winners after 2009, so the later period is thinly covered.

```python
total_per_year = df.groupby("Year").size()
cov = pd.DataFrame({"Total_records": total_per_year,
                    "With_age": aged.groupby("Year").size()}).fillna(0).astype(int)
cov["Coverage_%"] = (cov["With_age"] / cov["Total_records"] * 100).round(1)
cov = cov.reset_index()
cov.to_csv(f"{OUT_DIR}/age_coverage_by_year.csv", index=False)
cov.tail(10)
```

## 7. Plots

```python
# Age distribution per award and period
fig = px.box(aged, x="Award", y="AgeAtAward", color="Period", points="outliers",
             category_orders={"Award": AWARDS},
             title=f"Age at award by award type ({P1_LABEL} vs {P2_LABEL})",
             labels={"AgeAtAward": "Age at award"})
fig.write_html(f"{OUT_DIR}/age_boxplot.html"); fig.show()

# Age-band distribution (all awards pooled)
b = age_bins[age_bins.Award == "All awards"]
fig = px.bar(b, x="AgeBand", y="Share_%", color="Period", barmode="group",
             category_orders={"AgeBand": BIN_LABELS},
             title="Age-band distribution of winners (all awards pooled)",
             labels={"Share_%": "Share of winners in period (%)"})
fig.write_html(f"{OUT_DIR}/age_bands.html"); fig.show()

# Median age per award and period
fig = px.bar(age_sum, x="Award", y="median", color="Period", barmode="group", text="median",
             category_orders={"Award": AWARDS + ["All awards"]},
             title="Median age at award", labels={"median": "Median age"})
fig.write_html(f"{OUT_DIR}/age_median.html"); fig.show()

# Mean age by year
yr = aged.groupby("Year", as_index=False)["AgeAtAward"].mean()
fig = px.line(yr, x="Year", y="AgeAtAward", markers=True,
              title="Mean age at award by year (only years with ages)",
              labels={"AgeAtAward": "Mean age"})
fig.write_html(f"{OUT_DIR}/age_mean_by_year.html"); fig.show()

# Coverage
fig = px.bar(cov, x="Year", y=["With_age", "Total_records"], barmode="overlay",
             title="Winners with a known age vs all winners, by year",
             labels={"value": "Winners", "variable": ""})
fig.write_html(f"{OUT_DIR}/age_coverage.html"); fig.show()
```
