```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load the dataset from your local machine
# Replace 'param_vir_chakra_cleaned.csv' with your absolute path if the file is in a different folder
file_path = 'param_vir_chakra_cleaned.csv'
df = pd.read_csv(file_path)

# 2. Compute the X-axis metric:
# AgeAtAward represents the age at the time of receiving the award (or age at death for posthumous awards)
df['Plot_Age'] = df['AgeAtAward']

# 3. Set up the Seaborn and Matplotlib visualization style
sns.set_theme(style='whitegrid')
plt.figure(figsize=(12, 6))

# Create a combined histogram and kernel density estimation (KDE) plot
ax = sns.histplot(
    data=df,
    x='Plot_Age',
    bins=range(int(df['Plot_Age'].min()), int(df['Plot_Age'].max()) + 2),
    kde=True,
    color='#1f77b4',
    edgecolor='black',
    alpha=0.6,
)

# Customize titles and labels
plt.title(
    'Distribution of Param Vir Chakra Recipients\n(Age at Award / Age at Death if Posthumous)',
    fontsize=14,
    fontweight='bold',
    pad=15,
)
plt.xlabel(
    'Age (Years)\n[Age at Award, or Age at Death for Posthumous Awards]',
    fontsize=12,
    labelpad=10,
)
plt.ylabel('Number of People (Frequency)', fontsize=12, labelpad=10)

# Add gridlines and layout adjustments
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

# Save the plot locally (optional)
plt.savefig('param_vir_chakra_age_distribution.png', dpi=300)

# Display the plot window
plt.show()
```
