# Dataset Cleaning

### Required modules

```python
import pandas as pd
```

### Cleaning datasets function

This function does the following:
- Removes whitespaces
- Fixes ambiguous state/country spellings
- Make missing Birth Years and Years of award as NaN
- Finds Age when the recipient received the award



```python
def clean_dataframe(df):
    
    # Remove whitespaces 
    
    for col in df.select_dtypes(['object']).columns:
        df[col] = df[col].str.strip()
      
    if 'Field' in df.columns:  
        df['Field'] = (df['Field']
                .str.replace(r'\s*-\s*', '-', regex=True)   # "Others- X" -> "Others-X"
                .str.strip())
    
    
    # Standardized country/state name
    
    STATE_MAPPING = {
        'Panjab': 'Punjab',
        'Orissa': 'Odisha',
        'Pondicherry': 'Puducherry',
        'Uttranchal' : 'Uttarakhand',
        'Uttaranchal': 'Uttarakhand',
        'Chhatisgarh': 'Chhattisgarh',
        'Chattisgarh': 'Chhattisgarh',
        'Andra Pradesh': 'Andhra Pradesh',
        'Jharkhandh': 'Jharkhand',
        'J&K': 'Jammu and Kashmir',
        'Tamilnadu': 'Tamil Nadu',
        'Maharastra' : 'Maharashtra',
        'USA': 'United States of America',
        'U.S.A.' : 'United States of America',
        'United States': 'United States of America',
        'America': 'United States of America',
        'UAE': 'United Arab Emirates',
        'U.A.E.': 'United Arab Emirates',
        'UK': 'United Kingdom',
        'U.K.': 'United Kingdom',
        'Beirut': 'Lebanon',
        'Foreigner': 'Tibet'
    }
    
    # Clean the state/country column
    if 'State/Country' in df.columns:
        df['State/Country'] = (
            df['State/Country']
            .str.replace(r'\s*&\s*', ' and ', regex=True)
            .replace(STATE_MAPPING)
        )
        
    # Replacing missing values with NaN
    if 'Year' in df.columns:
        df['Year'] = pd.to_numeric(df['Year'], errors='coerce').astype('Int64')
        
    if 'Year of Birth' in df.columns:
        df['Year of Birth'] = pd.to_numeric(df['Year of Birth'], errors='coerce').astype('Int64')
        #Adding the Age at Award Column
        df['AgeAtAward'] = df['Year'] - df['Year of Birth']
        
    return df
```

### Importing datasets

```python
df_ashoka_chakra = pd.read_csv("../processed_datasets/ashoka_chakra_with_yob.csv")
df_national_awards = pd.read_csv("../processed_datasets/national_awards_with_yob.csv")
df_param_vir_chakra = pd.read_csv("../processed_datasets/param_vir_chakra_with_yob.csv")
```

### Cleaning and exporting

```python
df_ashoka_chakra = clean_dataframe(df_ashoka_chakra)
df_national_awards = clean_dataframe(df_national_awards)
df_param_vir_chakra = clean_dataframe(df_param_vir_chakra)

df_ashoka_chakra.to_csv('ashoka_chakra_cleaned.csv', index=False)
df_national_awards.to_csv('national_awards_cleaned.csv', index=False)
df_param_vir_chakra.to_csv('param_vir_chakra_cleaned.csv', index=False)
```
